```python
import numpy as np
import pandas as pd 
from matplotlib import pyplot as plt
```

```python
data = {
    "Names"  : ["Nandini" , "Niya" , "Lakshay" , "Nishka" , "Raunak"],
    "Maths"  : [34,56,23,89,56],
    "Science":[67,54,32,35,24],
    "English":[45,33,23,64,24]}

   
    
    
```

```python
df = pd.DataFrame(data)
```

```python
print("---THE STUDENT DATA---")
df
```

```python
df["total"] = df["Maths"] + df["Science"] + df["English"]
```

```python
df["Percentage"] = df["total"]/3
```

```python
def grade(p):
    if   p>=90:
        return "A"
    elif p>=75:
        return "B"
    elif p>=50:
        return "C"
    else:  
        return "D"    
    
```

```python
df["Grade"] = df["Percentage"].apply(grade)
```

```python
df["Rank"] = df["Percentage"].rank(ascending = False , method ="dense" ).astype(int)
```

```python
average_percentage = np.mean(df["Percentage"])
highest_percentage = np.max(df["Percentage"])
lowest_percentage = np.min(df["Percentage"])
topper = df.loc[df["Percentage"].idxmax(), "Names"]
lowest_scorer = df.loc[df["Percentage"].idxmin(), "Names"]

    
```

```python
print("---STUDENT REPORT---")
print("AVERAGE PERCENTAGE = " ,  round(average_percentage, 2))
print("HIGHEST PERCENTAGE = " ,  round( highest_percentage, 2))
print("LOWEST PERCENTAGE = "  ,  round(lowest_percentage , 2))
print("TOPPER = " , topper)
print("LOWEST SCORER= ", lowest_scorer)
```

```python
subavg_math = df["Maths"].mean()
subavg_english=df["English"].mean() 
subavg_science=df["Science"].mean()
print("---SUBJECT AVERAGES---")
print("MATHS = " , subavg_math)
print("ENGLISH = " , subavg_english)
print("SCIENCE = " , subavg_science)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(df["Names"], df["Percentage"])

plt.title("Student Percentage Analysis")
plt.xlabel("Students")
plt.ylabel("Percentage")
plt.show()
```

```python
subavg = [ subavg_math , subavg_english , subavg_science]
subjects = ["Math" , "English" , "Science " ] 
plt.pie(subavg , labels = subjects , autopct = "%1.1f%%")
plt.title("SUBJECT WISE AVERAGE ")
plt.show()
```

```python
plt.figure(figsize = (8,5))
plt.plot(df["Names"] , df["Maths"] , marker = "o" , label = "Maths")
plt.plot(df["Names"] , df["English"] , marker = "o" , label = "English")
plt.plot(df["Names"] , df["Science"] , marker = "o" , label = "Science")
plt.title("subject wise comparison")
plt.xlabel("students")
plt.ylabel("marks")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Percentage"], bins=5, edgecolor="black")
plt.title("Marks Percentage Distribution")
plt.xlabel("Percentage")
plt.ylabel("Number of Students")
plt.show()
```
